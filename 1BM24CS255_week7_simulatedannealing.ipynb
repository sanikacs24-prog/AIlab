```python

import random
import math

def calculate_conflicts(board, n):
    conflicts = 0

    for i in range(n):
        for j in range(i + 1, n):
            # Check diagonal conflicts
            if abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def simulated_annealing(n):
    # Random initial state (one queen per column)
    board = [random.randint(0, n - 1) for _ in range(n)]

    temperature = 1000
    cooling_rate = 0.99
    min_temperature = 0.01

    current_conflicts = calculate_conflicts(board, n)

    while temperature > min_temperature:

        if current_conflicts == 0:
            return board

        # Generate a neighboring state
        new_board = board.copy()

        col = random.randint(0, n - 1)
        new_board[col] = random.randint(0, n - 1)

        new_conflicts = calculate_conflicts(new_board, n)

        # Calculate change in conflicts
        delta = new_conflicts - current_conflicts

        # Accept better states or sometimes worse states
        if delta < 0:
            board = new_board
            current_conflicts = new_conflicts

        else:
            probability = math.exp(-delta / temperature)

            if random.random() < probability:
                board = new_board
                current_conflicts = new_conflicts

        # Reduce temperature
        temperature *= cooling_rate

    if current_conflicts == 0:
        return board

    return None


# Main program
n = int(input("Enter number of queens: "))

solution = simulated_annealing(n)

if solution:
    print("\nSolution found:")

    for row in range(n):
        for col in range(n):
            if solution[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()

    print("Board representation:", solution)

else:
    print("No solution found. Try again.")
```
