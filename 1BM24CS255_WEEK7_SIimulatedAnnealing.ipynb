```python

import math
import random

# Objective function
def objective(x):
    return x * x

# Simulated Annealing algorithm
def simulated_annealing():
    current = random.uniform(-10, 10)
    best = current

    temperature = 1000
    cooling_rate = 0.95
    min_temperature = 0.001

    while temperature > min_temperature:

        # Generate a neighboring solution
        new = current + random.uniform(-1, 1)

        # Calculate change in energy
        delta = objective(new) - objective(current)

        # Accept or reject the new solution
        if delta < 0:
            current = new
        else:
            probability = math.exp(-delta / temperature)

            if random.random() < probability:
                current = new

        # Update best solution
        if objective(current) < objective(best):
            best = current

        # Reduce temperature
        temperature *= cooling_rate

    return best, objective(best)


# Execute the algorithm
solution, value = simulated_annealing()

print("Best solution:", solution)
print("Minimum value:", value)
```
