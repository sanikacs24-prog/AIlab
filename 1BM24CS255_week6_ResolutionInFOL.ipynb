```python

def unify(x, y):
    if x == y:
        return True
    if x.islower():
        return True
    if y.islower():
        return True
    return False


def resolution(c1, c2):
    for l1 in c1:
        for l2 in c2:

            neg1 = l1.startswith("~")
            neg2 = l2.startswith("~")

            p1 = l1[1:] if neg1 else l1
            p2 = l2[1:] if neg2 else l2

            if p1.split("(")[0] == p2.split("(")[0] and neg1 != neg2:

                a1 = p1[p1.find("(")+1:p1.find(")")]
                a2 = p2[p2.find("(")+1:p2.find(")")]

                if unify(a1, a2):
                    new_clause = [x for x in c1 if x != l1]
                    new_clause += [x for x in c2 if x != l2]

                    if len(new_clause) == 0:
                        return "Success"

                    return new_clause

    return "No Resolution"


# Given clauses
C1 = ["~Human(x)", "Mortal(x)"]
C2 = ["Human(Socrates)"]
C3 = ["~Mortal(Socrates)"]

print("Clause 1:", C1)
print("Clause 2:", C2)
print("Clause 3:", C3)

# First resolution
R1 = resolution(C1, C2)
print("Resolvent 1:", R1)

# Second resolution
R2 = resolution(R1, C3)
print("Resolvent 2:", R2)

if R2 == "Success":
    print("Goal proved successfully!")
else:
    print("Resolution failed")
```
